```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv('/content/netflix_titles.csv')

print("Dataset Loaded Successfully!")
```

```python
missing_percentage = (
    data.isnull().sum() / len(data)
) * 100

print(missing_percentage.sort_values(ascending=False))
```

```python
data_clean = data.copy()

data_clean['director'] = data_clean['director'].fillna('Unknown')
data_clean['cast'] = data_clean['cast'].fillna('Unknown')
data_clean['country'] = data_clean['country'].fillna('Unknown')
data_clean['rating'] = data_clean['rating'].fillna('Not Rated')
data_clean['duration'] = data_clean['duration'].fillna('Unknown')

data_clean['date_added'] = data_clean['date_added'].fillna('Unknown')

print(data_clean.isnull().sum())
```

```python
print("Duplicate Rows:", data_clean.duplicated().sum())
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=data_clean, x='type')

plt.title('Movies vs TV Shows on Netflix')
plt.xlabel('Content Type')
plt.ylabel('Number of Titles')

plt.show()
```

```python
plt.figure(figsize=(12, 5))

data_clean['release_year'].value_counts().sort_index().plot()

plt.title('Netflix Content by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')

plt.show()
```

```python
top_countries = (
    data_clean[data_clean['country'] != 'Unknown']['country']
    .str.split(', ')
    .explode()
    .value_counts()
    .head(10)
)

plt.figure(figsize=(10, 5))

sns.barplot(x=top_countries.values, y=top_countries.index)

plt.title('Top 10 Countries by Netflix Titles')
plt.xlabel('Number of Titles')
plt.ylabel('Country')

plt.show()
```

```python
plt.figure(figsize=(12, 5))

data_clean['rating'].value_counts().plot(kind='bar')

plt.title('Netflix Content Ratings')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')

plt.xticks(rotation=45)
plt.show()
```

```python
movies = data_clean[data_clean['type'] == 'Movie'].copy()

movies['duration_minutes'] = pd.to_numeric(
    movies['duration'].str.replace(' min', '', regex=False),
    errors='coerce'
)

plt.figure(figsize=(10, 5))

sns.histplot(
    movies['duration_minutes'].dropna(),
    bins=30,
    kde=True
)

plt.title('Netflix Movie Duration Distribution')
plt.xlabel('Duration (Minutes)')
plt.ylabel('Number of Movies')

plt.show()
```

```python
top_genres = (
    data_clean['listed_in']
    .str.split(', ')
    .explode()
    .value_counts()
    .head(10)
)

plt.figure(figsize=(10, 6))

sns.barplot(x=top_genres.values, y=top_genres.index)

plt.title('Top 10 Netflix Genres')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')

plt.show()
```

```python
date_data = data_clean[
    data_clean['date_added'] != 'Unknown'
].copy()

date_data['date_added'] = pd.to_datetime(
    date_data['date_added'].str.strip(),
    errors='coerce'
)

date_data['year_added'] = date_data['date_added'].dt.year

date_data['year_added'].value_counts().sort_index()
```

```python
top_directors = (
    data_clean[data_clean['director'] != 'Unknown']['director']
    .str.split(', ')
    .explode()
    .value_counts()
    .head(10)
)

top_directors
```

```python
country_type = data_clean[
    data_clean['country'] != 'Unknown'
].copy()

country_type['country'] = country_type['country'].str.split(', ')

country_type = country_type.explode('country')

top_5 = country_type['country'].value_counts().head(5).index

plt.figure(figsize=(12, 6))

sns.countplot(
    data=country_type[country_type['country'].isin(top_5)],
    x='country',
    hue='type'
)

plt.title('Movies vs TV Shows in Top 5 Countries')
plt.xlabel('Country')
plt.ylabel('Number of Titles')

plt.show()
```
