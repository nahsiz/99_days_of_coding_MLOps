```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('/content/diamonds.csv')

print("Dataset Loaded Successfully!")
```

```python
print("Cut:", data['cut'].unique())
print("Color:", data['color'].unique())
print("Clarity:", data['clarity'].unique())
```

```python
print("Cut Distribution:")
print(data['cut'].value_counts())

print("\nColor Distribution:")
print(data['color'].value_counts())

print("\nClarity Distribution:")
print(data['clarity'].value_counts())
```

```python
data.isnull().sum()
```

```python
print("Duplicate Rows:", data.duplicated().sum())
```

```python
data_clean = data.drop_duplicates().copy()

print("Original Shape:", data.shape)
print("Cleaned Shape:", data_clean.shape)
print("Remaining Duplicates:", data_clean.duplicated().sum())
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(data_clean['price'], bins=30, kde=True)

plt.title('Diamond Price Distribution')
plt.xlabel('Price')
plt.ylabel('Frequency')

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(data_clean['carat'], bins=30, kde=True)

plt.title('Diamond Carat Distribution')
plt.xlabel('Carat')
plt.ylabel('Frequency')

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.countplot(data=data_clean, x='cut')

plt.title('Diamond Cut Distribution')
plt.xlabel('Cut Quality')
plt.ylabel('Count')

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(data=data_clean, x='cut', y='price')

plt.title('Diamond Price by Cut Quality')
plt.xlabel('Cut')
plt.ylabel('Price')

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.scatterplot(
    data=data_clean,
    x='carat',
    y='price',
    alpha=0.4
)

plt.title('Diamond Price vs Carat')
plt.xlabel('Carat')
plt.ylabel('Price')

plt.show()
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    data_clean.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Diamond Features Correlation')

plt.show()
```
